```python
import soccerdata as sd

fbref = sd.FBref(leagues = "GER-Bundesliga", seasons = [2122, 2223, 2324, 2425, 2526])
```

```python
# This scrapes team level match stats 
team_match_stats = fbref.read_team_match_stats()
team_match_stats.to_csv("fbref_team_match_stats.csv", index=False)
```

```python
# This scrapes team level season stats
team_season_stats = fbref.read_team_season_stats()
team_season_stats = team_season_stats.reset_index()

team_season_stats.columns = [
    "_".join(str(x) for x in col if x).strip("_")
    if isinstance(col, tuple)
    else col
    for col in team_season_stats.columns
]

team_season_stats.columns = (
    team_season_stats.columns
    .str.lower()
    .str.replace(" ", "_")
)

team_season_stats.to_csv("fbref_team_season_stats.csv", index=False)
```
