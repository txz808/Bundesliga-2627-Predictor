```python
import soccerdata as sd

understat = sd.Understat(leagues = "GER-Bundesliga", seasons = [2122, 2223, 2324, 2425, 2526], no_cache = True)
```

```python
# This scrapes team-level match data from Understat (xG and other advanced stats)
team_match_stats = understat.read_team_match_stats()
team_match_stats.to_csv("understat_team_match_stats.csv", index=False)
```
