```python
# This standardises team names so datasets from both sources are joined reliably
def normalise_team(x):
    if not isinstance(x, str):
        return x
    if " / " in x:
        parts = [old_to_new.get(p, p) for p in x.split(" / ")]
        seen = list(dict.fromkeys(parts))
        return " / ".join(seen)
    return old_to_new.get(x, x)

def is_fully_normalised(x):
    if not isinstance(x, str):
        return True
    return all(p in TEAM_NAME_MAP for p in x.split(" / "))

TEAM_NAME_MAP = {
    'Arminia Bielefeld': ['Arminia Bielefeld', 'Arminia'],
    'Augsburg': ['Augsburg'],
    'Bayer Leverkusen': ['Bayer Leverkusen', 'Leverkusen'],
    'Bayern Munich': ['Bayern Munich'],
    'Bochum': ['Bochum'],
    'Borussia Dortmund': ['Borussia Dortmund', 'Dortmund'],
    'Borussia Monchengladbach': ['Borussia M.Gladbach', 'Gladbach', 'Borussia Monchengladbach', "M'gladbach", 'Monchengladbach', 'Mönchengladbach'],
    'Darmstadt': ['Darmstadt', 'Darmstadt 98'],
    'Eintracht Frankfurt': ['Eintracht Frankfurt', 'Frankfurt'],
    'Freiburg': ['Freiburg'],
    'Greuther Furth': ['Greuther Fuerth', 'Greuther Fürth', 'Greuther Furth'],
    'Hamburger SV': ['Hamburger SV'],
    'Heidenheim': ['FC Heidenheim', 'Heidenheim'],
    'Hertha Berlin': ['Hertha Berlin', 'Hertha BSC'],
    'Hoffenheim': ['Hoffenheim'],
    'Holstein Kiel': ['Holstein Kiel'],
    'Koln': ['FC Cologne', 'FC Koln', 'Köln', 'Koln', '1. FC Koln'],
    'Mainz': ['Mainz 05', 'Mainz'],
    'RB Leipzig': ['RasenBallsport Leipzig', 'RB Leipzig', 'Rb Leipzig'],
    'Saint Pauli': ['St. Pauli', 'St Pauli', 'FC St. Pauli', 'Saint Pauli'],
    'Schalke': ['Schalke 04', 'Schalke'],
    'Stuttgart': ['VfB Stuttgart', 'Stuttgart'],
    'Union Berlin': ['Union Berlin'],
    'Werder Bremen': ['Werder Bremen'],
    'Wolfsburg': ['Wolfsburg'],
}

old_to_new= {old: new for new, olds in TEAM_NAME_MAP.items() for old in olds}
```

```python
# This loads the cleaned FBref team match statistics
fb_team_match_stats = pd.read_csv("/Users/tirenioladimeji/Documents/Coursework/bundesliga-2627-predictor/fbref/fbref_team_match_stats.csv")
understat_team_match_stats = pd.read_csv("/Users/tirenioladimeji/Documents/Coursework/bundesliga-2627-predictor/understat/understat_team_match_stats.csv")

dfs = {
    "fb_team_match_stats": fb_team_match_stats,
    "understat_team_match_stats": understat_team_match_stats,
}

for name, df in dfs.items():
    for col in ("team", "opponent", "home_team", "away_team"):
        if col in df.columns:
            df[col] = df[col].apply(normalise_team)

paths = {
    "understat_team_match_stats": "/Users/tirenioladimeji/Documents/Coursework/bundesliga-2627-predictor/understat/understat_team_match_stats.csv",
}
for name, df in dfs.items():
    df.to_csv(paths[name], index=False)
    print(f"Saved {name}")
```
